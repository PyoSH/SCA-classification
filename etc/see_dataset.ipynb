```python
import src.train_utils as utils
import src.dataset as dataset
import numpy as np

import sys
# npy_path = sys.path("..","data", "dataset_100ms_old", "audio_test_240814.npy")
npy_path = "/Users/seunghyunpyo/PycharmProjects/rnn_followup/data/dataset_100ms_old/audio_train_241004.npy"
dataset_npy = np.load(npy_path)

labels = dataset_npy[:,-1]
print(labels.shape)

has_nan = np.isnan(dataset_npy).any()

if has_nan:
    print("\n[결과] 🔴 데이터에 NaN 값이 포함되어 있습니다.")
    
    # 3. (선택 사항) NaN 값의 총 개수 세기
    nan_count = np.sum(np.isnan(dataset_npy))
    print(f"  총 NaN 개수: {nan_count} 개")
    
    # 4. (선택 사항) NaN 값이 포함된 행(row)의 개수 확인
    # axis=1 : 각 행(row)을 기준으로 검사
    rows_with_nan = np.isnan(dataset_npy).any(axis=1)
    print(f"  NaN이 포함된 행(샘플)의 수: {np.sum(rows_with_nan)} / {dataset_npy.shape[0]} 개")
    
else:
    print("\n[결과] 🟢 데이터에 NaN 값이 없습니다.")
```
